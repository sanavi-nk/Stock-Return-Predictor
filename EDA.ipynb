#### Downloading the datas

```python
import os

def get_price_data(ticker, start="2015-01-01", end="2025-01-01"):
    filepath = f"{ticker}_data.csv"
    if os.path.exists(filepath):
        df = pd.read_csv(filepath, index_col="Date", parse_dates=True)
        print(f"Loaded {ticker} from disk")
    else:
        df = yf.download(ticker, start=start, end=end)
        df.to_csv(filepath)
        print(f"Downloaded {ticker} from Yahoo Finance")
    return df[["Close", "Volume"]].copy()
```

```python
df = get_price_data("AAPL")
print(df.shape)
print(df.tail())
```

```python
import matplotlib.pyplot as plt

df["Close"].plot(title="AAPL Adjusted Close Price", figsize=(12, 4))
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.show()
```
